**1. Generate 10,000 random samples using built-in functions.**

```python
import numpy as np
from scipy import stats

np.random.seed(42)

sample_array = np.random.randint(low = 0, high = 100, size = 10000)

sample_list = sample_array.tolist()

print("Sample List : ", sample_list)
print("Mean : ",np.mean(sample_list))
print("Median : ", np.median(sample_list))
print("Standard Deviation : ",np.std(sample_list))

Mode = stats.mode(sample_list, keepdims = True)
mode_value = Mode.mode.item()
print("Mode : ", mode_value)
```
